```python
import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns


# Load model comparison results

results = pd.read_csv(
    "../reports/model_comparison.csv"
)


# Display results

results
```

```python
print("Model Comparison Results")

print(
    results.to_string(index=False)
)
```

```python
plt.figure(figsize=(8, 5))


sns.barplot(

    data=results,

    x="Algorithm",

    y="Accuracy"

)


plt.title(
    "Accuracy Comparison of ML Algorithms"
)


plt.xlabel(
    "Machine Learning Algorithm"
)


plt.ylabel(
    "Accuracy"
)


plt.xticks(
    rotation=15
)


plt.tight_layout()


plt.show()
```

```python
import os


os.makedirs(

    "../reports/evaluation",

    exist_ok=True

)


plt.figure(figsize=(8, 5))


sns.barplot(

    data=results,

    x="Algorithm",

    y="Accuracy"

)


plt.title(
    "Accuracy Comparison of ML Algorithms"
)


plt.xlabel(
    "Machine Learning Algorithm"
)


plt.ylabel(
    "Accuracy"
)


plt.xticks(
    rotation=15
)


plt.tight_layout()


plt.savefig(

    "../reports/evaluation/accuracy_comparison.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()
```

```python
metrics = [

    "Precision",

    "Recall",

    "F1 Score"

]


results_melted = results.melt(

    id_vars="Algorithm",

    value_vars=metrics,

    var_name="Metric",

    value_name="Score"

)


plt.figure(figsize=(10, 6))


sns.barplot(

    data=results_melted,

    x="Algorithm",

    y="Score",

    hue="Metric"

)


plt.title(
    "Precision, Recall and F1-score Comparison"
)


plt.xlabel(
    "Machine Learning Algorithm"
)


plt.ylabel(
    "Score"
)


plt.xticks(
    rotation=15
)


plt.tight_layout()


plt.show()
```

```python
plt.figure(figsize=(10, 6))


sns.barplot(

    data=results_melted,

    x="Algorithm",

    y="Score",

    hue="Metric"

)


plt.title(
    "Precision, Recall and F1-score Comparison"
)


plt.xlabel(
    "Machine Learning Algorithm"
)


plt.ylabel(
    "Score"
)


plt.xticks(
    rotation=15
)


plt.tight_layout()


plt.savefig(

    "../reports/evaluation/precision_recall_f1_comparison.png",

    dpi=300,

    bbox_inches="tight"

)


plt.show()
```

```python
best_algorithm = results.loc[

    results["F1 Score"].idxmax()

]


print("Best Performing Algorithm:")


print(
    best_algorithm["Algorithm"]
)


print("\nBest F1 Score:")


print(
    best_algorithm["F1 Score"]
)
```

```python
evaluation_summary = results.sort_values(

    by="F1 Score",

    ascending=False

)


evaluation_summary.to_csv(

    "../reports/evaluation/evaluation_summary.csv",

    index=False

)


print("Evaluation Summary Saved Successfully")
```

```python
import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns


df = pd.read_csv(
    "../data/processed/reviews_cleaned.csv"
)

df.head()
```

```python
print("Dataset Shape:")

print(df.shape)


print("\nMissing Values:")

print(df.isnull().sum())


print("\nDuplicate Reviews:")

print(df.duplicated().sum())


print("\nLabel Distribution:")

print(df["label"].value_counts())
```

```python
df["label"].value_counts().plot(
    kind="bar"
)

plt.title(
    "Fake vs Genuine Review Distribution"
)

plt.xlabel(
    "Review Type"
)

plt.ylabel(
    "Number of Reviews"
)

plt.xticks(
    [0, 1],
    ["Genuine", "Fake"],
    rotation=0
)

plt.show()
```

```python
df["review_length"] = (
    df["review_text"]
    .str.split()
    .str.len()
)

print(
    df["review_length"].describe()
)
```

```python
sns.boxplot(

    x="label",

    y="review_length",

    data=df

)

plt.title(
    "Review Length Comparison"
)

plt.xlabel(
    "Review Type"
)

plt.ylabel(
    "Number of Words"
)

plt.show()
```
